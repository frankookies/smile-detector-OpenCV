```python
from datetime import datetime
import cv2
import os

frame_count = 0
save_interval = 30  
save_dir = "smiles"


# importing the haarcascade files for face and smile detection
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
smile_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_smile.xml')

# enabling the webcam
webcam = cv2.VideoCapture(0)

# reads the frames from the webcam
while True:
    successful_frame_read, frame = webcam.read()
    if not successful_frame_read:
        print("Failed to get frame.")
        break

    # converting webcam frames to grayscale for easier face and smile detection
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # detecting faces in grayscale frame
    faces = face_cascade.detectMultiScale(gray_frame, scaleFactor=1.3, minNeighbors=5, minSize=(30, 30))

     # green rectangle around each face detected
    for (x, y, w, h) in faces:
        pass
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), thickness=2 )
        # extracting face region and detecting smiles
        face_gray = gray_frame[y:y+h, x:x+w]
        face_colour = frame[y:y+h, x:x+w]
        smiles = smile_cascade.detectMultiScale(face_gray, scaleFactor=1.7, minNeighbors=35, minSize=(50, 50))
        is_smiling = len(smiles) > 0

        # labeling smiling faces
        for (sx, sy, sw, sh) in smiles:
            cv2.rectangle(frame, (x+sx, y+sy), (x+sx+sw, y+sy+sh), (0, 0, 255), 2)
        lbl = "Smile :)" if is_smiling else "No Smile"
        cv2.putText(frame, lbl, (x, y-10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

        #Uncomment the following lines to save smiling face images at specified intervals
        """ 
        # creates a smiles directory if it doesn't exist  
        if not os.path.exists(save_dir):
            os.makedirs(save_dir)

        # saves smiling face images with timestamped filenames
        if is_smiling and frame_count % save_interval == 0:
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
            filename = f"smiles/smile_{timestamp}.jpg"
            cv2.imwrite(filename, face_colour)
        frame_count += 1  
        """

    cv2.imshow('Face and Smile Detection', frame)
    if cv2.waitKey(20) & 0xFF == ord('q'):
        break

webcam.release()
cv2.destroyAllWindows()
```
